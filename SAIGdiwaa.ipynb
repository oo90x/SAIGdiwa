## Time

เริ่มจาก time ขอดู format ปัจจุบันก่อน

````python
df['time'].head(10)
````

จะสังเกตุได้ว่า ['time'] อยู่ใน format {Year-Month-Date XX:XX} <br>
ปี, เดือน, วันที่ อาจจะส่งผลในแง่ของการเป็นวันเทศกาล แต่เราไม่ได้มีข้อมูลเยอะพอจะ pick up pattern จากตรงนั้นได้ เพราะฉะนั้นจะ drop ทิ้งไปเหลือแต่เวลาก่อน

````python
df['time'] = pd.to_datetime(df['time'])
df['hour'] = df['time'].dt.hour

df = df.drop(columns=['time'])
hour_col = df.pop('hour')
df.insert(0, 'hour', hour_col)
df.head(20)
````

````python
df['hour'].value_counts()
````

จัดกลุ่มข้อมูลเวลาให้เป็นกลุ่ม ๆ ได้แก่ Dawn, Morning, Afternoon, Dusk, Night

````python
def time_period(hour):

  '''
  change number in col['hour'] into time period
  return: str
  '''

  if 4 <= hour <= 6: # ตี 4 ถึง 6 โมง
      return 'Dawn'
  elif 6 < hour < 12: # 7 โมง ถึง 11 โมง
      return 'Morning'
  elif 12 <= hour < 17: # เที่ยง ถึง 4 โมง
      return 'Afternoon'
  elif 17 <= hour < 19: # 5 โมง ถึง 6 โมง
      return 'Dusk'
  else:  # หนึ่งทุ่ม ถึงก่อน ตี 4
      return 'Night'
````

````python
df['time_period'] = df['hour'].apply(time_period)
df = df.drop(columns=['hour'])
time_period_col = df.pop('time_period')
df.insert(0, 'time_period', time_period_col)
df.head(20)
````

## Road_characteristic

ต่อมาดูที่ road_characteristic

````python
df['road_characteristic'].value_counts().head(40)
````

มีข้อมูลที่ยังไม่ clean เรียบร้อยเช่น ทางสามแยก (Y) กับ ทางสามแยก(Y) ซึ่งมีความหมายแบบเดียวกัน แต่มีการเว้นวรรคทำให้กลายเป็นคนละ features ตรงนี้ต้องทำการ clean ก่อน

````python
df['road_characteristic'].unique()
````

````python
replace_map = {
    'ทางสามแยก(T)': 'ทางสามแยก (T)',
    'ทางสามแยก(Y)': 'ทางสามแยก (Y)',
    'ทางแยกต่างระดับ/Ramps': 'ทางแยกต่างระดับ / Ramps',
    'สะพาน (สะพานข้ามแม่น้ำ/ข้ามคลอง)': 'สะพาน',
    'เข้าพื้นที่สาธารณะหรือเชิงพานิชย์': 'ทางเชื่อมเข้าพื้นที่สาธารณะ/เชิงพาณิชย์',
}

df['road_characteristic'] = df['road_characteristic'].replace(replace_map)
````

แก้ชื่อของ ทางแยกรูปตัว + เพราะจะมีปัญหาตอนแยก col ด้วย strip

````python
df['road_characteristic'] = df['road_characteristic'].str.replace(
    'ทางแยกรูปตัว +', 'ทางแยกรูปตัวบวก', regex=False
)
````

````python
df['road_characteristic'].unique()
````

จะสังเกตุได้ว่าข้อมูลบาง class ไม่ใช่ class ใหม่แต่เป็นการรวมกันของสองข้อมูล ตรงนี้จะ deal ด้วยการแยก column ออกมา แล้วสำหรับตัวที่มี class เดียวจะเกิด NaN ให้ fill ด้วย ไม่ระบุ แทน

````python
split_cols = df['road_characteristic'].str.split('+', n=1, expand=True)
df['road_type'] = split_cols[0].str.strip()
df['road_slope'] = split_cols[1].str.strip()

df['road_slope'] = df['road_slope'].fillna('ไม่ระบุ')
````

## Fatalities & Injuries = Severity

ข้อมูลมี imbalance ชัดเจน กรณีบาดเจ็บและเสียชีวิตมากมีแค่ 1 เคส ทำให้โมเดลมีโอกาสไม่เรียนรู้ หรือ ถ้าหากอยู่ใน test set ก็จะวัดผลไม่ค่อยได้ ตอนแรกจะทำการคำนวณเป็นตัวเลขขึ้นมา แต่จะทำให้มีหลายค่าเกินไป และมีบางตัวเลขที่จะเป็น outlier ค่อนข้างเยอะ คิดว่าท่า regression ผมยังเล่นยากไป อาจจะลองท่า classification เลยจะใช้อีกวิธี ก็คือการจัดกลุ่มแทน

ผมเลยสร้าง function ชื่อว่า calc_severity ขึ้นมา เพื่อจัดกลุ่มออกมาเป็น 7 กลุ่มตามความรุนแรง (พยายามจัดกลุ่มให้ละเอียดที่สุดเพื่อไม่ให้เสีย insight แล้ว แต่ว่าแค่นี้ก็ยังทำให้แต่ละ class เกิด imbalance เลยน่าจะแบ่งได้มากสุดแค่นี้ แล้วก็แก้ไขปัญหาด้วยการ ใช้ split, class_weight, metric ที่เหมาะสมแทน)

````python
def calc_severity(row):
    f = row['fatalities']
    i = row['injuries']

    if f == 0 and i == 0:
        return 'No Casualty'
    elif f == 0 and 1 <= i <= 2:
        return 'Minor Injury'
    elif f == 0 and i >= 3:
        return 'Serious Injury (Multiple)'
    elif f == 1 and i == 0:
        return 'Single Fatality'
    elif f == 1 and i >= 1:
        return 'Single Fatality + Injuries'
    elif 2 <= f <= 4:
        return 'Multiple Fatalities (2-4)'
    elif f >= 5:
        return 'Mass Casualty (5+ Fatalities)'
    else:
        return 'Other'

df['severity'] = df.apply(calc_severity, axis=1)
df['severity'].value_counts()
````

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy.stats import chi2_contingency
from itertools import combinations

def cramers_v(x, y):
    confusion_matrix = pd.crosstab(x, y)
    chi2 = chi2_contingency(confusion_matrix)[0]
    n = confusion_matrix.sum().sum()
    phi2 = chi2 / n
    r, k = confusion_matrix.shape
    # bias correction (Bergsma 2013)
    phi2corr = max(0, phi2 - ((k-1)*(r-1)) / (n-1))
    rcorr = r - ((r-1)**2) / (n-1)
    kcorr = k - ((k-1)**2) / (n-1)
    return np.sqrt(phi2corr / min((kcorr-1), (rcorr-1)))

cols = df.columns.tolist()
cramers_matrix = pd.DataFrame(np.zeros((len(cols), len(cols))), index=cols, columns=cols)

for col1, col2 in combinations(cols, 2):
    v = cramers_v(df[col1], df[col2])
    cramers_matrix.loc[col1, col2] = v
    cramers_matrix.loc[col2, col1] = v

np.fill_diagonal(cramers_matrix.values, 1.0)

plt.figure(figsize=(10, 8))
sns.heatmap(cramers_matrix, annot=True, fmt=".2f", cmap="coolwarm",
            vmin=0, vmax=1, square=True, linewidths=0.5)
plt.title("Cramér's V Association Between Categorical Features")
plt.tight_layout()
plt.show()
````
